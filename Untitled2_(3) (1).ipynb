1.	timestamp  - время лога (@timestamp)
2.	metadata  - данные о данных (служебное Elastic поле)
3.	log_name  - имя лога (например, Sysmon/Operational)

5.	type  - тип API парсера (wineventlog - современный, eventlogging - старый)
6.	record_number  - номер записи в журнале Windows
7.	computer_name  - имя рабочей машины, где реально произошло событие
8.	version  - версия схемы самого события (задается провайдером, не версия ОС)
9.	event_data  - словарь с деталями (процессы, cmdline, IP)
10.	source_name  - имя провайдера, создавшего событие (например, Microsoft-Windows-Sysmon)
11.	thread_id  - ID потока (TID), сгенерировавшего событие
12.	user  - пользователь (словарь с SID и именем)
13.	provider_guid  - GUID провайдера-источника (не хэш процесса, а ID компонента Windows)
14.	host  - хост, который физически собрал лог (коллектор, например WECserver)
15.	beat  - инфа об агенте сбора (Winlogbeat, его версия и хост)
16.	opcode  - код операции внутри задачи (часто просто Info)
17.	event_id  - ID события Windows (1, 3, 4688 и т.д.)
18.	task  - категория события (например, Process Creation)
19.	process_id  - PID процесса, сгенерировавшего лог
20.	level  - уровень важности (Information, Warning, Error)
21.	keywords  - флаги аудита (например, [Audit Success])
22.	activity_id  - GUID для связки событий в одну логическую операцию
23.	user_data  - XML-данные события (часто пустые, если есть event_data)

## 2. Очистка данных и обогащение

Мы удалим "мусорные" колонки, распакуем словарь `user` на удобные `user_name` и `user_sid`, а также извлечем ключевые поля из `event_data` (командные строки, имена процессов, IP-адреса) в отдельные колонки для быстрого поиска.

```python
# Удаляем только технические поля, не нужные для основного анализа
drop_cols = [
    '@metadata',
    'beat',
    'host',
    'type'
]

df = df.drop(columns=drop_cols, errors='ignore')

print("Удалены:", drop_cols)
print("Оставшиеся колонки:", df.columns.tolist())
print("Размер df:", df.shape)
```

```python
def extract_user_info(user_dict):
    if isinstance(user_dict, dict):
        return pd.Series({
            'user_name': user_dict.get('name'),
            'user_sid': user_dict.get('identifier'),
            'user_domain': user_dict.get('domain')
        })

    return pd.Series({
        'user_name': None,
        'user_sid': None,
        'user_domain': None
    })

df[['user_name', 'user_sid', 'user_domain']] = (
    df['user'].apply(extract_user_info)
)

df = df.drop(columns=['user'])
```

```python
def extract_features(row):
    ev = row['event_data']

    if not isinstance(ev, dict):
        return pd.Series({
            'proc_name': None,
            'source_image': None,
            'target_img': None,
            'cmd_line': None,
            'src_ip': None,
            'dest_ip': None,
            'granted_access': None
        })

    return pd.Series({
        # Имя процесса, если оно присутствует в событии
        'proc_name': ev.get('Image'),

        # Отдельно сохраняем источник и цель доступа
        'source_image': ev.get('SourceImage'),
        'target_img': ev.get('TargetImage'),

        # Командная строка
        'cmd_line': ev.get('CommandLine'),

        # IP-адреса: учитываем разные названия полей
        'src_ip': ev.get('SourceIp') or ev.get('IpAddress'),
        'dest_ip': ev.get('DestinationIp'),

        # Права доступа к объекту
        'granted_access': ev.get('GrantedAccess')
    })


features = df.apply(extract_features, axis=1)
df = pd.concat([df, features], axis=1)

print("Количество заполненных значений:")
print(df[[
    'proc_name', 'source_image', 'target_img',
    'cmd_line', 'src_ip', 'dest_ip'
]].notna().sum())
```

В событиях PowerShell наиболее часто встречаются Event ID 800 (18 366 записей) и 4103 (15 632 записи). Они связаны с выполнением PowerShell pipeline и регистрацией активности модулей. Таким образом, распределение Event ID позволило выделить несколько приоритетных направлений дальнейшего исследования: доступ к чувствительным процессам, изменения реестра, загрузка библиотек, сетевые соединения, изменение привилегий и выполнение команд PowerShell.

При этом высокая частота определённого события сама по себе не свидетельствует о вредоносной активности. Для выявления возможных признаков компрометации необходимо исследовать содержимое событий, связанные процессы, учётные записи, временные отметки и взаимосвязи между разными типами событий.


1. Почему изначально мы подумали, что это атака? (Подозрения)

Первичный вывод о возможной компрометации строился на классической сигнатуре техники MITRE ATT&CK T1546.008 (Accessibility Features).
• Аномальный флаг /debug: Запуск системной утилиты с аргументом utilman.exe /debug выглядит крайне подозрительно. В хакерской практике флаг /debug часто ассоциируется со скрытой отладкой или внедрением кода (Image File Execution Options — IFEO). Когда злоумышленник подменяет диспетчер утилит на свою консоль (cmd.exe), в логах запуск этой скрытой консоли часто отображается именно с флагами отладки.

Любая нестандартная активность на нем (тем более с правами SYSTEM на экране блокировки) автоматически расценивается как инцидент максимального приоритета (Critical Alert).
• Совпадение по времени: Запуск произошел в 23:44, что идеально укладывалось в таймлайн общего всплеска активности логов в сети (пик атаки по вашему минутному графику шел как раз в 23:42–23:43). Мы резонно предположили, что хакер закрепился в системе.

2. Почему это на 100% НЕ атака? (Опровержение)

Когда мы применили метод Process Tree Analysis (анализ дерева процессов) и посмотрели на события, произошедшие сразу «после», гипотеза атаки полностью рассыпалась.
• Легитимный родитель: Родителем процесса выступил winlogon.exe. Это доказывает, что вызов утилиты произошел штатно: кто-то нажал кнопку специальных возможностей (или комбинацию клавиш) на физическом экране приветствия Windows.
• Отсутствие вредоносных дочерних процессов: Если бы бэкдор был активен, вместо или внутри utilman.exe запустилась бы командная строка хакера (cmd.exe, powershell.exe или агент Empire). Однако ни одного детекта запуска консоли от этого PID не зафиксировано.

```python
# Находим самое первое и самое последнее событие в датасете
min_time = df['@timestamp'].min()
max_time = df['@timestamp'].max()
total_duration = max_time - min_time

print(f"=== Анализ временных рамок датасета ===")
print(f"Начало логов (Первое событие): {min_time}")
print(f"Конец логов (Последнее событие):  {max_time}")
print(f"Общая продолжительность логов:   {total_duration}")
```

```python
import matplotlib.pyplot as plt

# 1. Фильтруем данные строго за день атаки (14 мая 2019 года)
df_attack_day = df[
    (df['@timestamp'] >= '2019-05-14 00:00:00+00:00') &
    (df['@timestamp'] <= '2019-05-14 23:59:59+00:00')
].copy()

# 2. Группируем с более точным шагом в 15 минут
attack_day_trend = df_attack_day.groupby([
    pd.Grouper(key='@timestamp', freq='15min'),
    'computer_name'
]).size().unstack(fill_value=0)

# 3. Строим график
plt.figure(figsize=(14, 6))
attack_day_trend.plot(ax=plt.gca(), linewidth=2)

# 4. Оформление
plt.title('Детальный таймлайн атаки: Активность серверов 14 мая 2019 года (шаг — 15 мин)', fontsize=14, pad=15)
plt.xlabel('Время (14 мая)', fontsize=12)
plt.ylabel('Количество событий за 15 минут', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Имя компьютера', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Фильтруем логи строго для хоста HR001 в промежутке между 15:00 и 22:00
afternoon_logs = df[
    (df['computer_name'] == 'HR001.shire.com') &
    (df['@timestamp'] >= '2019-05-14 15:00:00+00:00') &
    (df['@timestamp'] <= '2019-05-14 22:00:00+00:00')
]

print(f"Всего логов за этот период: {len(afternoon_logs)}")

# Смотрим, какие именно Event ID генерировали этот фон
print("\n=== Распределение типов событий (Event ID) с 15:00 до 22:00 ===")
print(afternoon_logs.groupby(['source_name', 'event_id']).size().reset_index(name='Количество').sort_values('Количество', ascending=False))
```
