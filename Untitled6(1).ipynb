```python
from scapy.all import PcapReader, IP, TCP, UDP
import pandas as pd
import ipaddress

path = '/content/drive/MyDrive/Нейросети/Friday-WorkingHours.pcap'
limit = 100000

data = []

with PcapReader(path) as pcap:
    cnt = 0
    for pkt in pcap:
        if cnt >= limit:
            break
        if IP in pkt:
            s_ip = pkt[IP].src
            d_ip = pkt[IP].dst

            s_port = 0
            d_port = 0
            flags = ''

            if TCP in pkt:
                s_port = pkt[TCP].sport
                d_port = pkt[TCP].dport
                flags = str(pkt[TCP].flags)
            elif UDP in pkt:
                s_port = pkt[UDP].sport
                d_port = pkt[UDP].dport

            # Определяем внешний/внутренний IP (Пункт 2)
            try:
                is_priv_s = ipaddress.ip_address(s_ip).is_private
                is_priv_d = ipaddress.ip_address(d_ip).is_private
            except:
                is_priv_s = False
                is_priv_d = False

            data.append({
                'time': float(pkt.time),
                's_ip': s_ip,
                'd_ip': d_ip,
                's_mac': pkt.src,
                'd_mac': pkt.dst,
                's_port': s_port,
                'd_port': d_port,
                'proto': pkt[IP].proto,
                'ttl': pkt[IP].ttl,
                'flags': flags,
                'len': len(pkt),
                's_is_private': is_priv_s,
                'd_is_private': is_priv_d
            })
        cnt += 1

df = pd.DataFrame(data)

# Интерпретация пункт 4
print("Интерпретация результатов")
print(f"Загружено {len(df)} пакетов.")
print(f"Из них от внутренних адресов: {df['s_is_private'].sum()}")
print(f"Из них от внешних адресов: {(~df['s_is_private']).sum()}")
print("Вывод: Большая часть трафика инициируется изнутри (внутренние адреса), что логично для рабочей станции в офисе.")
df.head()
```

```python
from scapy.all import conf
import ipaddress

SPECIAL_MACS = {
    "ff:ff:ff:ff:ff:ff": "широковещательный", "33:33:00:00:00:01": "IPv6 mcast", "01:00:5e:00:00:01": "IPv4 mcast",
    "01:80:c2:00:00:00": "STP", "01:80:c2:00:00:02": "STP", "01:80:c2:00:00:0e": "LLDP",
    "01:80:c2:00:00:03": "802.1x", "01:80:c2:00:00:08": "STP", "01:80:c2:00:00:10": "LLDP",
    "01:80:c2:00:00:21": "PROFINET", "00:00:0c:07:ac:00": "VRRP", "00:00:0c:07:ac:01": "VRRP",
    "00:00:5e:00:01:01": "VRRPv4", "00:00:5e:00:02:01": "VRRPv6", "01:00:5e:00:00:05": "OSPF",
    "01:00:5e:00:00:06": "OSPF", "01:00:5e:00:00:12": "VRRP", "01:00:5e:00:00:13": "PIM",
    "01:00:5e:00:00:0a": "EIGRP", "01:00:5e:00:00:0d": "PIM", "01:00:5e:7f:ff:fa": "mDNS",
    "01:00:5e:7f:ff:fb": "mDNS", "01:00:5e:00:00:fb": "mDNS"
}

GROUP_V4 = {
    "224.0.0.1": "все узлы (IPv4)", "224.0.0.2": "все маршрутизаторы", "224.0.0.5": "OSPF",
    "224.0.0.9": "RIPv2", "224.0.0.10": "EIGRP", "224.0.0.13": "PIM", "224.0.0.22": "IGMPv3",
    "224.0.0.251": "mDNS", "224.0.0.252": "LLMNR", "239.255.255.250": "SSDP"
}

def mcast_mac(group):
    a = ipaddress.ip_address(group)
    if a.version == 4:
        low = int(a) & 0x7fffff
        return "01:00:5e:" + ".".join(format((low >> s) & 0xff, "02x") for s in (16, 8, 0))
```

```python
MAC_NAME = {mcast_mac(g): f"{g} ({n})" for g, n in GROUP_V4.items()}

def classify_mac(mac):
    mac = mac.lower()
    best = None
    for p in SPECIAL_MACS:
        if mac.startswith(p):
            if best is None or len(p) > len(best):
                best = p
    if best is not None:
        return "служебный", SPECIAL_MACS[best]
    if mac in MAC_NAME:
        return "multicast", MAC_NAME[mac]
    try:
        if int(mac.split(':')[0], 16) & 2:
            return "локально администрируемый", "-"
    except:
        pass
    return "обычный", "-"
```

1.1. Таблица служебных MAC-адресов назначения
Смотрим, какие служебные адреса встречаются в трафике и в каком количестве. Это помогает понять, есть ли в сети активный широковещательный шум.

```python
def special_dst_table(df):
    g = df.groupby('d_mac')
    tab = g.agg(количество=('len', 'size')).reset_index()

    cats = []
    protos = []

    for i, row in tab.iterrows():
        c, p = classify_mac(row['d_mac'])
        cats.append(c)
        protos.append(p)

    tab['категория'] = cats
    tab['протокол'] = protos
    return tab.sort_values('количество', ascending=False).head(10)

special_dst_table(df)
```

1.2. Инвентаризация MAC-адресов (источников)
Определяем производителей сетевых карт по первым трем байтам (OUI). Это нужно для восстановления топологии и понимания того, какие устройства находятся в сети (компьютеры, принтеры, сетевое оборудование).

```python
def vendor(mac):
    if mac is None:
        return None
    try:
        v = conf.manufdb._get_manuf(mac)
    except Exception:
        return None
    return None if (v is None or v.lower() == mac.lower()) else v

VENDOR_CLASS = {
    "сетевое оборудование": ["cisco", "juniper", "palo alto", "fortinet", "huawei", "routerboard", "mikrotik"],
    "АСУ ТП": ["siemens", "moxa", "phoenix", "schneider", "rockwell", "abb", "beckhoff", "omron", "yokogawa", "honeywell"],
    "виртуализация": ["vmware", "xensource", "qemu", "parallels", "virtualbox"],
    "ТК и серверы": ["dell", "hewlett", "lenovo", "intel", "supermicro", "gigabyte", "asrock", "asus", "micro-star"],
    "потребительские": ["apple", "samsung", "xiaomi", "raspberry", "ubiquiti", "tp-link", "d-link", "google"]
}

def vendor_class(v):
    if not v:
        return None
    lv = v.lower()
    for c, keys in VENDOR_CLASS.items():
        for k in keys:
            if k in lv:
                return c
    return "прочее"
def mac_inventory(df):
    s = df[df['s_mac'].notna()]
    inv = s.groupby('s_mac').agg(количество=('len', 'size')).reset_index()

    ips = []
    n_ips = []
    vendors = []
    classes = []

    for i, row in inv.iterrows():
        mac = row['s_mac']
        ip_list = list(set(s[s['s_mac'] == mac]['s_ip'].dropna()))
        ips.append(sorted(ip_list)[:4])
        n_ips.append(len(ip_list))

        v = vendor(mac)
        vendors.append(v)
        classes.append(vendor_class(v))

    inv['ip'] = ips
    inv['n_ip'] = n_ips
    inv['производитель'] = vendors
    inv['класс'] = classes

    return inv.sort_values('количество', ascending=False).head(10)

mac_inventory(df)
```

Блок 2. Восстановление топологии (поиск маршрутизаторов)
Эвристика простая: если один MAC-адрес имеет несколько разных IP-адресов, значит через него проходит трафик разных устройств. Скорее всего, это интерфейс маршрутизатора или шлюза.

```python

def is_group_ip(ip):
    if not isinstance(ip, str):
        return False
    try:
        a = ipaddress.ip_address(ip)
        return a.is_multicast or ip == "255.255.255.255" or ip.endswith(".255")
    except:
        return False

def ip_mac_pairs(df):
    mac_list = []
    ip_list = []

    for idx, row in df.iterrows():
        if row['s_mac'] != 'ff:ff:ff:ff:ff:ff' and not is_group_ip(row['s_ip']):
            mac_list.append(row['s_mac'])
            ip_list.append(row['s_ip'])

        if row['d_mac'] != 'ff:ff:ff:ff:ff:ff' and not is_group_ip(row['d_ip']):
            mac_list.append(row['d_mac'])
            ip_list.append(row['d_ip'])

    res = pd.DataFrame({'mac': mac_list, 'ip': ip_list})
    return res.drop_duplicates().reset_index(drop=True)

pairs = ip_mac_pairs(df)
pairs.head()
```

2.1. Поиск потенциальных маршрутизаторов
Группируем данные по MAC-адресу. Если у одного MAC-адреса больше одного уникального IP, мы помечаем его как потенциальный роутер. Также выведем производителя этого устройства.

```python
from scapy.all import conf

def vendor(mac):
    if mac is None:
        return None
    try:
        v = conf.manufdb._get_manuf(mac)
    except Exception:
        return None
    return None if (v is None or v.lower() == mac.lower()) else v

def find_routers(pairs, min_ips=2):
    rows = []
    for mac, group in pairs.groupby('mac'):
        ips = set(group['ip'])
        if len(ips) < min_ips:
            continue

        own = []
        for ip in ips:
            try:
                if ipaddress.ip_address(ip).is_private:
                    own.append(ip)
            except:
                pass

        v = vendor(mac)

        rows.append({
            'mac': mac,
            'всего_ip': len(ips),
            'частные_ip': ", ".join(sorted(own)) if own else "-",
            'производитель': v if v else "-"
        })

    return pd.DataFrame(rows).sort_values('всего_ip', ascending=False).reset_index(drop=True)

routers = find_routers(pairs)
routers.head(10)
```

2.2. Схема сети (Визуализация)
Построим граф связей между устройствами (MAC) и их IP-адресами. Чтобы ноутбук не завис при отрисовке, возьмем только первые 150 связей из нашей таблицы pairs.

```python
import networkx as nx
import matplotlib.pyplot as plt
import ipaddress

def build_topology(pairs, routers):
    G = nx.Graph()
    small_pairs = pairs.head(50)

    router_macs = list(routers['mac'])

    for idx, row in small_pairs.iterrows():
        m = row['mac']
        ip = row['ip']

        kind = "external"
        if ipaddress.ip_address(ip).is_private:
            kind = "host"

        if m in router_macs:
            kind = "router"

        G.add_node(m, kind="mac", label=m)
        G.add_node(ip, kind=kind, label=ip)
        G.add_edge(m, ip)

    return G
```

```python
def draw_topology(G, title, ax):
    pos = nx.spring_layout(G, seed=42, k=0.8)

    nx.draw_networkx_edges(G, pos, ax=ax, alpha=0.3, edge_color="gray")

    styles = {
        "mac": ("o", "#7287bd", 150),
        "router": ("D", "#e0834e", 400),
        "host": ("o", "#a8e6a4", 200),
        "external": ("h", "#9a9a9a", 200)
    }

    for kind, (shape, color, size) in styles.items():
        nodes = [n for n, d in G.nodes(data=True) if d.get('kind') == kind]
        nx.draw_networkx_nodes(G, pos, nodelist=nodes, node_shape=shape, node_color=color, node_size=size, ax=ax)

    labels = {n: d.get('label', n) for n, d in G.nodes(data=True)}
    nx.draw_networkx_labels(G, pos, labels, font_size=6, ax=ax)

    ax.set_title(title)
    ax.axis("off")

G = build_topology(pairs, routers)
fig, ax = plt.subplots(figsize=(12, 8))
draw_topology(G, "Схема связей MAC и IP (50 связей)", ax)
plt.tight_layout()
plt.show()
```

```python
def get_subnet(ip):
    # Получаем подсеть /24 для проверки разных сегментов
    try:
        return str(ipaddress.ip_network(str(ip) + '/24', strict=False))
    except:
        return "unknown"

def find_routers_advanced(pairs):
    rows = []
    for mac, group in pairs.groupby('mac'):
        ips = set(group['ip'])
        if len(ips) < 2:
            continue

        # Проверяем, есть ли IP из разных подсетей (Пункт 3)
        subnets = set()
        for ip in ips:
            subnets.add(get_subnet(ip))

        own = [ip for ip in ips if ipaddress.ip_address(ip).is_private]

        rows.append({
            'mac': mac,
            'всего_ip': len(ips),
            'уникальных_подсетей': len(subnets),
            'частные_ip': ", ".join(sorted(own)) if own else "-",
        })

    res = pd.DataFrame(rows)
    return res.sort_values('уникальных_подсетей', ascending=False).reset_index(drop=True)

routers = find_routers_advanced(pairs)


print(f"Найдено {len(routers)} потенциальных роутеров.")
print("Роутер — это устройство, которое имеет IP-адреса в разных подсетях (сегментах).")
print("Смотрим таблицу: если у MAC уникальных подсетей > 1, значит он маршрутизирует трафик между сегментами.")
routers.head(10)
```

Блок 3. Сетевой уровень: внутренние и внешние адреса, обогащение
В этом блоке мы классифицируем IP-адреса (глобальные, частные, multicast и т.д.), проанализируем TTL для выявления NAT, а также обогатим внешние адреса данными о странах, автономных системах (AS) и репутации.

```python
import ipaddress
import pandas as pd

def ip_kind(ip):
    a = ipaddress.ip_address(ip)
    if a.version == 4 and ip == "255.255.255.255":
        return "broadcast"
    if a.is_unspecified:
        return "unspecified (0.0.0.0 / ::)"
    if a.is_loopback:
        return "loopback"
    if a.is_multicast:
        return "multicast"
    if a.is_link_local:
        return "link-local (APIPA)"
    if a.is_private:
        if not (a.version == 4 and ip.endswith(".255")):
            return "частный (RFC 1918 / ULA)"
        else:
            return "broadcast подсети"
    if a.is_reserved:
        return "зарезервированный"
    return "глобальный"

def ip_kinds(df):
    ips = set(df['s_ip'].dropna()) | set(df['d_ip'].dropna())
    s = pd.Series(sorted(ips))
    return s.map(ip_kind).value_counts().rename("адресов")

print(ip_kinds(df))
```

3.1. Анализ TTL и выявление NAT
Анализируем TTL (Time To Live) у пакетов, идущих от внешних адресов. Если TTL у всех внешних пакетов строго одинаковый, это верный признак того, что перед нами работает NAT или прокси, который переписывает заголовки пакетов.

```python
priv_cnt = 0
ext_cnt = 0
ttl_ext = {}

for i, row in df.iterrows():
    kind = ip_kind(row['s_ip'])
    if "частный" in kind:
        priv_cnt += 1
    else:
        ext_cnt += 1
        t = int(row['ttl'])
        if t not in ttl_ext:
            ttl_ext[t] = 0
        ttl_ext[t] += 1

print("Пакетов от внутренних адресов:", priv_cnt)
print("Пакетов от внешних адресов:", ext_cnt)
print("Распределение TTL у внешних:", ttl_ext)
```

Блок 4. Транспортный уровень: сборка сессий
На этом этапе мы применяем наивный алгоритм сборки TCP/UDP сессий. Направление определяется сравнением портов: меньший порт считается портом сервера. Все пакеты с одинаковым набором (клиент, сервер, порт сервера, протокол) объединяются в одну сессию.

```python
import numpy as np
import pandas as pd

def naive_sessions(df):
    p = df[['time', 's_ip', 'd_ip', 's_port', 'd_port', 'len', 'proto']].copy()
    p['s_port'] = p['s_port'].astype(int)
    p['d_port'] = p['d_port'].astype(int)

    swap = (p['s_port'] > p['d_port']) | ((p['s_port'] == p['d_port']) & (p['s_ip'] < p['d_ip']))

    p['srv_ip'] = np.where(swap, p['d_ip'], p['s_ip'])
    p['cli_ip'] = np.where(swap, p['s_ip'], p['d_ip'])
    p['srv_port'] = np.where(swap, p['d_port'], p['s_port'])
    p['cli_port'] = np.where(swap, p['s_port'], p['d_port'])

    g = p.groupby(['cli_ip', 'srv_ip', 'srv_port', 'proto'])

    s = g.agg(starts=('time', 'min'), ends=('time', 'max'), pkts=('len', 'size'), bytes=('len', 'sum')).reset_index()

    return s.sort_values('starts').reset_index(drop=True)
```

```python
def build_sessions_syn(df):
    sessions = []

    # Проходим по каждой уникальной связке (IP1, порт1, IP2, порт2, протокол)
    # Чтобы не зависеть от направления, сортируем IP и порты
    grouped = df.groupby(['proto',
                          df[['s_ip', 'd_ip']].min(axis=1),
                          df[['s_ip', 'd_ip']].max(axis=1),
                          df[['s_port', 'd_port']].min(axis=1),
                          df[['s_port', 'd_port']].max(axis=1)])

    for name, group in grouped:
        proto = name[0]
        group = group.sort_values('time')

        initiator_ip = None
        initiator_port = None

        # Пункт 1: для TCP ищем SYN-пакет
        if proto == 6:
            syns = group[group['flags'].str.contains('S', na=False)]
            # Ищем именно чистой SYN (без ACK)
            pure_syns = syns[~syns['flags'].str.contains('A', na=False)]
            if len(pure_syns) > 0:
                first_syn = pure_syns.iloc[0]
                initiator_ip = first_syn['s_ip']
                initiator_port = first_syn['s_port']

        # Для UDP и других - просто первый пакет
        if initiator_ip is None:
            first_pkt = group.iloc[0]
            initiator_ip = first_pkt['s_ip']
            initiator_port = first_pkt['s_port']

        # Определяем отвечающую сторону
        first_pkt = group.iloc[0]
        if initiator_ip == first_pkt['s_ip']:
            target_ip = first_pkt['d_ip']
            target_port = first_pkt['d_port']
        else:
            target_ip = first_pkt['s_ip']
            target_port = first_pkt['s_port']

        sessions.append({
            'cli_ip': initiator_ip,
            'cli_port': initiator_port,
            'srv_ip': target_ip,
            'srv_port': target_port,
            'proto': proto,
            'starts': group['time'].min(),
            'ends': group['time'].max(),
            'pkts': len(group),
            'bytes': group['len'].sum(),
            'is_ext_srv': not ipaddress.ip_address(target_ip).is_private
        })

    return pd.DataFrame(sessions).sort_values('starts').reset_index(drop=True)

naive = build_sessions_syn(df)

# Интерпретация
print(f"Собрано {len(naive)} сессий с корректным определением инициатора.")
print(f"Сессий, где сервер находится во внешней сети: {naive['is_ext_srv'].sum()}")
print("Благодаря анализу SYN-флагов, мы точно знаем, кто начал соединение. Это критично для поиска вредоносной активности (например, если внутренний хост стучится на внешний сервер).")
naive.head()
```

Блок 5. Выявление подозрительного трафика по обогащённым сессиям

```python
from scapy.all import PcapReader, TCP, Raw, IP, DNS

zip_ips = set()
nx_ips = set()
tls_ips = set()

count = 0
with PcapReader(path) as reader:
    for p in reader:
        if count >= limit:
            break
        count += 1

        if p.haslayer(IP):
            s_ip = p[IP].src
            d_ip = p[IP].dst

            if p.haslayer(TCP) and p.haslayer(Raw):
                try:
                    payload = bytes(p[Raw].load)
                    payload_lower = payload.lower()

                    if b'zip' in payload_lower:
                        zip_ips.add(s_ip)
                        zip_ips.add(d_ip)

                    if p[TCP].dport == 443 or p[TCP].sport == 443:
                        if len(payload) >= 2 and payload[0] == 0x16 and payload[1] == 0x03:
                            tls_ips.add(s_ip)
                            tls_ips.add(d_ip)
                except:
                    pass

            if p.haslayer(DNS) and p[DNS].qr == 1:
                if p[DNS].rcode == 3:
                    nx_ips.add(d_ip)

print('IP с zip:', len(zip_ips))
print('IP с nxdomain:', len(nx_ips))
print('IP с шифрованием TLS:', len(tls_ips))
```

5.1. Начисление баллов подозрительности

```python
scores = []

for idx, row in naive.iterrows():
    score = 0

    if row['bytes'] > 100000:
        score += 10

    if row['cli_ip'] in zip_ips or row['srv_ip'] in zip_ips:
        score += 20

    if row['cli_ip'] in nx_ips or row['srv_ip'] in nx_ips:
        score += 15

    if row['srv_port'] == 443 and row['srv_ip'] not in tls_ips:
        score += 2

    scores.append(score)

naive['score'] = scores
```
